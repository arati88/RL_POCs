```python
import gymnasium as gym              #Gymnasium provides standardized RL environments and Handles state transitions, rewards, and episode logic
```

```python
# Create the Taxi environment
# This environment defines states, actions, and reward structure

env = gym.make("Taxi-v3")  # Initializes the Taxi RL environment and Defines State space, Action space, Reward structure

state, info = env.reset()       #Reset Environment: Starts a new episode
print("Initial state:", state)

total_reward = 0

for step in range(5):
    action = env.action_space.sample()      # Agent selects an action randomly, This simulates an untrained agent

    state, reward, terminated, truncated, info = env.step(action)     #Gymnasium returns five values: state: next state, reward: feedback from environment, terminated: task success or failure,truncated: time-limit reached, info: additional details
    total_reward += reward         # Tracks how well the agent is performing

    print(f"Step {step+1}")
    print("Action:", action)
    print("State:", state)
    print("Reward:", reward)

    if terminated or truncated:
        print("Episode ended")
        break

env.close()
print("Total reward:", total_reward)
```
