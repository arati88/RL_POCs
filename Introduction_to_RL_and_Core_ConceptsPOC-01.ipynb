```python
import gymnasium as gym              #Gymnasium provides standardized RL environments and Handles state transitions, rewards, and episode logic
```

```python
# Create the Taxi environment
# -----------------------------
# This environment defines:
# - States: 0 to 499 (each representing a taxi location, passenger location, and destination)
# - Actions: 6 discrete actions (move south, north, east, west, pickup, dropoff)
# - Rewards: based on successful pickup/dropoff or penalties for illegal moves

env = gym.make("Taxi-v3")  # Initializes the Taxi RL environment and Defines State space, Action space, Reward structure

# Reset the environment to start a new episode
# Returns:
# - state: initial state of the environment
# - info: optional information dictionary
state, info = env.reset()       #Reset Environment: Starts a new episode
print("Initial state:", state)

# Variable to track total cumulative reward in the episode
total_reward = 0

# Loop for a fixed number of steps (simulate an agent interacting with environment)
for step in range(5):

    # Agent selects a random action from the available action space
    # This simulates an untrained or exploratory agent
    action = env.action_space.sample()     

    # Execute the chosen action in the environment
    # Returns:
    # - state: next state after taking action
    # - reward: immediate reward from this action
    # - terminated: True if the task is successfully completed
    # - truncated: True if episode ends due to time limit
    # - info: additional diagnostic info
    state, reward, terminated, truncated, info = env.step(action)     

    # Update total cumulative reward
    total_reward += reward         # Tracks how well the agent is performing

     # Print details of the current step for tracking/debugging
    print(f"Step {step+1}")
    print("Action:", action)
    print("State:", state)
    print("Reward:", reward)

    # Stop the episode if it ended (success or time limit reached)
    if terminated or truncated:
        print("Episode ended")
        break

# Close the environment to free resources
env.close()
print("Total reward:", total_reward)
```
